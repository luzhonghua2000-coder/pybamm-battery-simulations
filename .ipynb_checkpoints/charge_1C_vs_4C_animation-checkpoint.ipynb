# 1C vs 4C Charge — Dynamic Li Concentration Visualisation**Fresh project. Self-contained. Run top to bottom.**## What this does1. Simulates a lithium-ion cell charging from **empty** at **1C** and at **4C**   using the DFN (Doyle-Fuller-Newman) model2. Displays, in a time loop:   - a **cursor moving along the voltage curve**   - **Li concentration across the cell** (anode | separator | cathode)   - **Li concentration inside an anode particle** (centre -> surface)   - **Li concentration inside a cathode particle** (centre -> surface)## Requirements```pip install "pybamm[plot]" jupyterlab```**Best practice: Kernel -> Restart Kernel and Run All Cells**

---## Step 1 — Imports and environment check

````python
import sysimport timeimport numpy as npimport matplotlib.pyplot as pltfrom matplotlib.patches import Patchfrom matplotlib.animation import FuncAnimation, PillowWriterfrom IPython.display import display, clear_output, HTMLimport pybammprint("Python :", sys.version.split()[0])print("PyBaMM :", pybamm.__version__)print("Kernel :", sys.executable)
````

---## Step 2 — Model, parameters, and cell geometry**DFN is required here.** Simpler models (SPM/SPMe) do not resolve electrolyteconcentration across the cell, which is half of what we want to see.`Chen2020` is a well-characterised parameter set for the LG M50 cell(NMC811 cathode, graphite-silicon anode).

````python
model = pybamm.lithium_ion.DFN()param = pybamm.ParameterValues("Chen2020")# --- cell geometry (metres) ---L_n   = param["Negative electrode thickness [m]"]   # anodeL_s   = param["Separator thickness [m]"]L_p   = param["Positive electrode thickness [m]"]   # cathodeL_tot = L_n + L_s + L_pR_n = param["Negative particle radius [m]"]R_p = param["Positive particle radius [m]"]print("CELL GEOMETRY")print(f"  Anode      : {L_n*1e6:7.2f} um   (0 -> {L_n*1e6:.1f})")print(f"  Separator  : {L_s*1e6:7.2f} um   ({L_n*1e6:.1f} -> {(L_n+L_s)*1e6:.1f})")print(f"  Cathode    : {L_p*1e6:7.2f} um   ({(L_n+L_s)*1e6:.1f} -> {L_tot*1e6:.1f})")print(f"  TOTAL      : {L_tot*1e6:7.2f} um")print()print("PARTICLES")print(f"  Anode particle radius   : {R_n*1e6:.3f} um")print(f"  Cathode particle radius : {R_p*1e6:.3f} um")# saturation limits - used as reference lines latertry:    CN_MAX = param["Maximum concentration in negative electrode [mol.m-3]"]    CP_MAX = param["Maximum concentration in positive electrode [mol.m-3]"]    print()    print(f"  Anode   c_max : {CN_MAX:9.1f} mol/m3")    print(f"  Cathode c_max : {CP_MAX:9.1f} mol/m3")except KeyError:    CN_MAX = CP_MAX = None    print("\n  (c_max parameter names differ in this version - skipping limit lines)")
````

---## Step 3 — Define the two charge protocolsEach run is a **CC-CV charge**:| Phase | What happens ||---|---|| **CC** (constant current) | Push a fixed current until voltage reaches 4.2 V || **CV** (constant voltage) | Hold 4.2 V while current tapers to C/50 |The only difference between the two runs is the CC rate: **1C vs 4C**.Both start from **empty** (`initial_soc=0`), so this is a pure charge comparison —no discharge preamble needed.

````python
V_MAX = 4.2   # upper cutoff voltagedef charge_protocol(rate):    return pybamm.Experiment([        f"Charge at {rate} until {V_MAX} V",     # CC phase        f"Hold at {V_MAX} V until C/50",         # CV phase    ])exp_1C = charge_protocol("1C")exp_4C = charge_protocol("4C")print("1C protocol:"); print(exp_1C)print()print("4C protocol:"); print(exp_4C)
````

---## Step 4 — Solve both simulationsThe 4C run is stiffer (steeper gradients) so it takes longer to solve.**Expect roughly 30-90 seconds total.** This is normal, not a hang.

````python
t0 = time.time()sim_1C = pybamm.Simulation(model, parameter_values=param, experiment=exp_1C)sol_1C = sim_1C.solve(initial_soc=0)print(f"1C solved in {time.time()-t0:5.1f} s  ->  {sol_1C.termination}")t1 = time.time()sim_4C = pybamm.Simulation(model, parameter_values=param, experiment=exp_4C)sol_4C = sim_4C.solve(initial_soc=0)print(f"4C solved in {time.time()-t1:5.1f} s  ->  {sol_4C.termination}")print(f"\nTotal: {time.time()-t0:.1f} s")
````

---## Step 5 — Verify the data before plotting**Always do this.** A `NaN` anywhere here will silently produce a blank plot later.

````python
sols   = [sol_1C, sol_4C]labels = ["1C charge", "4C charge"]colors = ["tab:blue", "tab:red"]def raw_arrays(sol):    """Pull flat float arrays straight out of the solution."""    t = np.asarray(sol["Time [h]"].entries, dtype=float).ravel()    v = np.asarray(sol["Voltage [V]"].entries, dtype=float).ravel()    i = np.asarray(sol["Current [A]"].entries, dtype=float).ravel()    return t, v, ifor sol, lab in zip(sols, labels):    t, v, i = raw_arrays(sol)    print(f"--- {lab} ---")    print(f"  samples   : {t.size}")    print(f"  time      : {t[0]:.4f} -> {t[-1]:.4f} h  ({t[-1]*60:.1f} min)")    print(f"  voltage   : {np.nanmin(v):.4f} -> {np.nanmax(v):.4f} V")    print(f"  current   : {np.nanmin(i):.4f} -> {np.nanmax(i):.4f} A")    print(f"  NaN check : t={np.isnan(t).sum()} v={np.isnan(v).sum()} i={np.isnan(i).sum()}")    print()t_1C = raw_arrays(sol_1C)[0][-1] * 60t_4C = raw_arrays(sol_4C)[0][-1] * 60print(f"CHARGE TIME   1C: {t_1C:6.1f} min")print(f"              4C: {t_4C:6.1f} min")print(f"Speed-up: {t_1C/t_4C:.2f}x   (NOT 4x - see the discussion at the end)")
````

### Static voltage checkIf this renders correctly, everything downstream will work.

````python
fig, ax = plt.subplots(figsize=(11, 5))for sol, lab, c in zip(sols, labels, colors):    t, v, i = raw_arrays(sol)    ax.plot(t * 60, v, color=c, lw=2, label=lab)    # mark where CC ends and CV begins (current starts tapering)    cc_end = np.argmax(v >= V_MAX - 1e-3)    if cc_end > 0:        ax.plot(t[cc_end] * 60, v[cc_end], "o", color=c, ms=9,                mec="black", zorder=5)        ax.annotate(f"CC->CV\n{t[cc_end]*60:.1f} min",                    (t[cc_end] * 60, v[cc_end]),                    textcoords="offset points", xytext=(8, -28),                    fontsize=8, color=c)ax.axhline(V_MAX, color="gray", ls="--", lw=1, alpha=0.7)ax.text(1, V_MAX + 0.01, f"{V_MAX} V cutoff", fontsize=8, color="gray")ax.set(xlabel="Time [min]", ylabel="Voltage [V]",       title="1C vs 4C charge - terminal voltage")ax.legend(); ax.grid(alpha=0.3)plt.tight_layout(); plt.show()
````

---## Step 6 — Precompute every frameEvaluating the solution inside the display loop makes it stutter, so we buildall arrays first.Two robustness details that matter:- **Voltage via `np.interp`** on the raw arrays — this cannot return `NaN`,  unlike calling the solution object exactly at a domain boundary- **`EPS` inset** when querying spatial variables, so we never ask for a time  value sitting exactly on the endpoint

````python
N_FRAMES = 120          # lower to 60 if precompute is slowEPS      = 1e-6t_end  = [raw_arrays(s)[0][-1] for s in sols]T_END  = max(t_end)times  = np.linspace(0.0, T_END, N_FRAMES)# spatial gridsx_grid = np.linspace(1e-9, L_tot - 1e-9, 150)   # across the cellr_n_g  = np.linspace(0.0, R_n, 40)              # into anode particler_p_g  = np.linspace(0.0, R_p, 40)              # into cathode particleframes = []for sol, tm in zip(sols, t_end):    t_arr, v_arr, i_arr = raw_arrays(sol)    tq   = np.clip(times, t_arr[0], t_arr[-1])    v_fr = np.interp(tq, t_arr, v_arr)          # NaN-free    i_fr = np.interp(tq, t_arr, i_arr)    ce, cn, cp = [], [], []    for t in times:        tc = min(max(t, t_arr[0] + EPS), tm - EPS)        ce.append(np.squeeze(            sol["Electrolyte concentration [mol.m-3]"](t=tc, x=x_grid)))        cn.append(np.squeeze(            sol["Negative particle concentration [mol.m-3]"](                t=tc, x=L_n / 2, r=r_n_g)))        cp.append(np.squeeze(            sol["Positive particle concentration [mol.m-3]"](                t=tc, x=L_n + L_s + L_p / 2, r=r_p_g)))    frames.append(dict(        t=t_arr, v=v_arr, i=i_arr,        v_fr=v_fr, i_fr=i_fr,        ce=np.array(ce, dtype=float),        cn=np.array(cn, dtype=float),        cp=np.array(cp, dtype=float),    ))# sanity checkok = Truefor f, lab in zip(frames, labels):    bad = sum(int(np.isnan(f[k]).sum()) for k in ["v_fr", "ce", "cn", "cp"])    print(f"{lab}: NaNs = {bad}  {'OK' if bad == 0 else '<-- PROBLEM'}")    ok &= (bad == 0)print(f"\nPrecomputed {N_FRAMES} frames over 0 -> {T_END*60:.1f} min")print("Ready to animate." if ok else "Fix NaNs before animating.")
````

---## Step 7 — Build the figureFour panels, created once. The loop in Step 8 only updates the line data,which is what keeps it smooth.

````python
def nan_safe_limits(arr, frac=0.06):    arr = np.asarray(arr, dtype=float)    lo, hi = np.nanmin(arr), np.nanmax(arr)    if not (np.isfinite(lo) and np.isfinite(hi)):        return (0.0, 1.0)    if hi - lo < 1e-12:        return (lo - 0.5, hi + 0.5)    m = (hi - lo) * frac    return (lo - m, hi + m)def build_figure():    fig, axes = plt.subplots(2, 2, figsize=(14, 9))    ax_v, ax_e = axes[0, 0], axes[0, 1]    ax_n, ax_p = axes[1, 0], axes[1, 1]    cursors, l_e, l_n, l_p = [], [], [], []    for f, lab, c in zip(frames, labels, colors):        ax_v.plot(f["t"] * 60, f["v"], color=c, lw=1.8, alpha=0.85, label=lab)        cursors.append(ax_v.plot([], [], "o", color=c, ms=14,                                 mec="black", mew=1.6, zorder=6)[0])        l_e.append(ax_e.plot([], [], color=c, lw=2.2, label=lab)[0])        l_n.append(ax_n.plot([], [], color=c, lw=2.2, label=lab)[0])        l_p.append(ax_p.plot([], [], color=c, lw=2.2, label=lab)[0])    vline = ax_v.axvline(0, color="k", ls="--", lw=1.2, alpha=0.7, zorder=5)    ax_v.axhline(V_MAX, color="gray", ls=":", lw=1, alpha=0.8)    # --- panel limits ---    ax_v.set(xlim=(0, T_END * 60),             ylim=nan_safe_limits(np.concatenate([f["v"] for f in frames])),             xlabel="Time [min]", ylabel="Voltage [V]",             title="Voltage  (dot = current time)")    ax_e.set(xlim=(0, L_tot * 1e6),             ylim=nan_safe_limits(np.concatenate([f["ce"].ravel() for f in frames])),             xlabel="Position across cell [um]",             ylabel="Electrolyte Li [mol/m3]",             title="Li in ELECTROLYTE across cell")    ax_n.set(xlim=(0, R_n * 1e6),             ylim=nan_safe_limits(np.concatenate([f["cn"].ravel() for f in frames])),             xlabel="centre  ->  surface  [um]", ylabel="Li [mol/m3]",             title="Li inside ANODE particle")    ax_p.set(xlim=(0, R_p * 1e6),             ylim=nan_safe_limits(np.concatenate([f["cp"].ravel() for f in frames])),             xlabel="centre  ->  surface  [um]", ylabel="Li [mol/m3]",             title="Li inside CATHODE particle")    # shade anode / separator / cathode    ax_e.axvspan(0, L_n * 1e6, alpha=0.12, color="tab:gray")    ax_e.axvspan(L_n * 1e6, (L_n + L_s) * 1e6, alpha=0.12, color="tab:orange")    ax_e.axvspan((L_n + L_s) * 1e6, L_tot * 1e6, alpha=0.12, color="tab:green")    yt = ax_e.get_ylim()[1]    ax_e.text(L_n * 1e6 / 2, yt * 0.985, "ANODE", ha="center", fontsize=8)    ax_e.text((L_n + L_s / 2) * 1e6, yt * 0.985, "SEP", ha="center", fontsize=8)    ax_e.text((L_n + L_s + L_p / 2) * 1e6, yt * 0.985, "CATHODE",              ha="center", fontsize=8)    # saturation reference lines    if CN_MAX is not None:        ax_n.axhline(CN_MAX, color="darkred", ls="--", lw=1.2, alpha=0.8)        ax_n.text(0.4, CN_MAX, " c_max (plating risk)", fontsize=8,                  color="darkred", va="bottom")        ax_n.set_ylim(top=max(ax_n.get_ylim()[1], CN_MAX * 1.05))    for a in (ax_v, ax_e, ax_n, ax_p):        a.legend(loc="best", fontsize=9)        a.grid(alpha=0.3)    suptitle = fig.suptitle("", fontsize=14)    fig.tight_layout()    return fig, (cursors, l_e, l_n, l_p, vline, suptitle)def set_frame(k, artists):    """Update every artist to frame k."""    cursors, l_e, l_n, l_p, vline, suptitle = artists    t = times[k]    for j, f in enumerate(frames):        cursors[j].set_data([min(t, t_end[j]) * 60], [f["v_fr"][k]])        l_e[j].set_data(x_grid * 1e6, f["ce"][k])        l_n[j].set_data(r_n_g * 1e6, f["cn"][k])        l_p[j].set_data(r_p_g * 1e6, f["cp"][k])    vline.set_xdata([t * 60, t * 60])    st = []    for f, lab in zip(frames, labels):        phase = "CV" if abs(f["v_fr"][k] - V_MAX) < 5e-3 else "CC"        st.append(f"{lab}: {f['v_fr'][k]:.3f} V, {abs(f['i_fr'][k]):.2f} A [{phase}]")    suptitle.set_text(f"t = {t*60:6.2f} min     |     " + "     |     ".join(st))    return cursors + l_e + l_n + l_p + [vline, suptitle]print("Figure builder ready.")
````

---## Step 8 — THE LOOP (dynamic display)An explicit `for` loop over time. Each iteration updates the four panels andredraws in place using `clear_output(wait=True)`.Adjust `PAUSE` to change playback speed. Interrupt the kernel (stop button)to stop early.

````python
PAUSE = 0.06     # seconds between frames - larger = slowerfig, artists = build_figure()for k in range(N_FRAMES):    set_frame(k, artists)    clear_output(wait=True)     # replace previous frame    display(fig)    time.sleep(PAUSE)print("Loop finished.")
````

### Loop over a chosen time windowSometimes you want to slow down and focus on the early CC phase, where thegradients build fastest.

````python
def loop_window(t_start_min, t_stop_min, pause=0.12):    """Replay only the frames between two times (in minutes)."""    ks = [k for k in range(N_FRAMES)          if t_start_min <= times[k] * 60 <= t_stop_min]    if not ks:        print("No frames in that window.")        return    fig, artists = build_figure()    for k in ks:        set_frame(k, artists)        clear_output(wait=True)        display(fig)        time.sleep(pause)    print(f"Replayed {len(ks)} frames, {t_start_min}-{t_stop_min} min")# first 10 minutes, slowly - this is where 4C diverges mostloop_window(0, 10, pause=0.15)
````

---## Step 9 — Player version (play / pause / scrub slider)The loop above runs once through. This gives you proper playback controls,and it is what you would export for sharing.

````python
fig2, artists2 = build_figure()anim = FuncAnimation(    fig2,    lambda k: set_frame(k, artists2),    frames=N_FRAMES,    interval=70,    blit=False,)plt.close(fig2)          # suppress the duplicate static figureHTML(anim.to_jshtml())   # play / pause / speed / scrub
````

---## Step 10 — Snapshot comparison at a fixed timeA static figure is often better than an animation for a report or slide.

````python
def snapshot(t_min):    k = int(np.argmin(np.abs(times * 60 - t_min)))    fig, ax = plt.subplots(1, 3, figsize=(16, 4.5))    for f, lab, c in zip(frames, labels, colors):        ax[0].plot(x_grid * 1e6, f["ce"][k], color=c, lw=2, label=lab)        ax[1].plot(r_n_g * 1e6,  f["cn"][k], color=c, lw=2, label=lab)        ax[2].plot(r_p_g * 1e6,  f["cp"][k], color=c, lw=2, label=lab)    ax[0].axvspan(0, L_n * 1e6, alpha=0.12, color="tab:gray")    ax[0].axvspan(L_n * 1e6, (L_n + L_s) * 1e6, alpha=0.12, color="tab:orange")    ax[0].axvspan((L_n + L_s) * 1e6, L_tot * 1e6, alpha=0.12, color="tab:green")    if CN_MAX is not None:        ax[1].axhline(CN_MAX, color="darkred", ls="--", lw=1.2, alpha=0.8)    ax[0].set(xlabel="Position across cell [um]",              ylabel="Electrolyte Li [mol/m3]", title="ELECTROLYTE across cell")    ax[1].set(xlabel="centre -> surface [um]", ylabel="Li [mol/m3]",              title="ANODE particle")    ax[2].set(xlabel="centre -> surface [um]", ylabel="Li [mol/m3]",              title="CATHODE particle")    for a in ax:        a.legend(); a.grid(alpha=0.3)    fig.suptitle(f"Snapshot at t = {times[k]*60:.2f} min", fontsize=13)    plt.tight_layout(); plt.show()for tm in [1, 5, 15]:    snapshot(tm)
````

---## How to read the plots### Voltage panelThe 4C curve rises almost vertically and slams into the 4.2 V cutoff within acouple of minutes. That vertical jump is **overpotential** — the extra voltageneeded to drive current through the cell's internal resistances. None of it isstored energy; it becomes heat.### Electrolyte panel (position across cell)At 1C the profile stays nearly flat: ion transport comfortably keeps up.At 4C a steep gradient develops — lithium is stripped from the electrolyte onone side and piles up on the other. This is **diffusion limitation**, and itis a hard physical ceiling. You cannot engineer around it by simply pushingmore current.### Anode particle panel — the important oneThe x-axis runs from the particle **centre (0)** to its **surface**.During charge, lithium arrives at the surface and must diffuse inward.At 4C, the surface concentration climbs steeply while the centre stays nearlyempty. Solid-state diffusion is too slow to redistribute it. When that surfacevalue approaches `c_max` (dashed red line), arriving lithium has **nowhere leftto intercalate** — so it deposits as metallic lithium on the surface instead.That is **lithium plating**: the dominant fast-charge degradation mechanism.It permanently consumes lithium, and the resulting dendrites are a genuinesafety hazard.### Cathode particle panelThe mirror image — lithium is leaving, so the surface **depletes** first whilethe centre stays full.---## Three points worth being able to say out loud**1. Fast charging is not proportionally faster.**Check the charge times printed in Step 5. 4C is roughly 4x the current butnowhere near 4x faster, because the cell hits the voltage limit early andspends most of the charge in the slow CV taper.**2. The bottleneck is transport, not capacity.**Both cells hold identical lithium. The 4C cell is limited by how fast lithiumcan *move* — through the electrolyte and into the particles.**3. Cells fail on local conditions, not averages.**A cell reading 50% SOC on average can already be plating lithium at the anodeparticle surfaces nearest the separator. Average state of charge tells younothing about whether damage is occurring. This is exactly why physics-basedmodels earn their keep over equivalent-circuit models.---## Things to try next| Experiment | Change ||---|---|| Find the safe charge limit | Add 2C, 3C, 6C and see where surface concentration hits `c_max` || Add heat | `pybamm.lithium_ion.DFN(options={"thermal": "lumped"})` || Model plating explicitly | `options={"lithium plating": "irreversible"}` || Compare chemistries | Swap `Chen2020` for another parameter set || Multi-stage fast charge | Step the rate down: 4C -> 2C -> 1C as voltage climbs |
