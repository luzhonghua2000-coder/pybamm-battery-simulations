```python
import matplotlib.animation as animation
import matplotlib.pyplot as plt
import numpy as np
import pybamm

# 1. SETUP MODEL & PARAMETERS
# ==========================================
print("Setting up PyBaMM simulations...")
model = pybamm.lithium_ion.DFN()  # Use full electrochemical model
params = pybamm.ParameterValues("Chen2020")  # Standard LG M50 NMC811/Graphite cell

# FIX: Set initial State of Charge (SoC) to 0 (fully discharged cell)
params["Initial SoC"] = 0

# Define experiments for 1C and 4C Constant Current (CC) charging
experiment_1C = pybamm.Experiment(["Charge at 1C until 4.2 V"])
experiment_4C = pybamm.Experiment(["Charge at 4C until 4.2 V"])

# Solve both models
sim_1C = pybamm.Simulation(model, parameter_values=params, experiment=experiment_1C)
sol_1C = sim_1C.solve()

sim_4C = pybamm.Simulation(model, parameter_values=params, experiment=experiment_4C)
sol_4C = sim_4C.solve()
# ==========================================
# 2. EXTRACT DATA & INTERPOLATE UNIFORM TIME
# ==========================================
# Common time array for animation (0 to 900 seconds)
t_max = min(sol_1C["Time [s]"].entries[-1], sol_4C["Time [s]"].entries[-1])
t_anim = np.linspace(0, t_max, 150)  # 150 frames

# Extract variables across time
v_1C = sol_1C["Terminal voltage [V]"](t_anim)
v_4C = sol_4C["Terminal voltage [V]"](t_anim)

# Spatial profiles (x domain: cell thickness)
x_ce = sol_1C["Electrolyte concentration [mol.m-3]"].domain_data["cell"]
ce_1C = sol_1C["Electrolyte concentration [mol.m-3]"](t=t_anim, x=x_ce)
ce_4C = sol_4C["Electrolyte concentration [mol.m-3]"](t=t_anim, x=x_ce)

# Particle profiles (r domain: particle radius at surface x=x_surface)
r_n = sol_1C[
    "Negative particle concentration [mol.m-3]"
].domain_data["negative particle"]
cs_n_1C = sol_1C["Negative particle concentration [mol.m-3]"](
    t=t_anim, r_n=r_n
)  # spatial average/surface
cs_n_4C = sol_4C["Negative particle concentration [mol.m-3]"](t=t_anim, r_n=r_n)

r_p = sol_1C[
    "Positive particle concentration [mol.m-3]"
].domain_data["positive particle"]
cs_p_1C = sol_1C["Positive particle concentration [mol.m-3]"](
    t=t_anim, r_p=r_p
)  # spatial average/surface
cs_p_4C = sol_4C["Positive particle concentration [mol.m-3]"](t=t_anim, r_p=r_p)

# Average out spatial x dependence for particle comparison across the particle radius 'r'
cs_n_1C_r = np.mean(cs_n_1C, axis=1)  # average across x dimension
cs_n_4C_r = np.mean(cs_n_4C, axis=1)

cs_p_1C_r = np.mean(cs_p_1C, axis=1)
cs_p_4C_r = np.mean(cs_p_4C, axis=1)

# ==========================================
# 3. BUILD DYNAMIC PLOTS
# ==========================================
fig, axs = plt.subplots(2, 2, figsize=(12, 8))
fig.suptitle(
    "Fast Charge Dynamic Simulation: 1C vs 4C Rate",
    fontsize=14,
    fontweight="bold",
)

# Panel 1: Voltage Curve
axs[0, 0].plot(
    sol_1C["Time [s]"].entries,
    sol_1C["Terminal voltage [V]"].entries,
    "b--",
    alpha=0.5,
    label="1C Full",
)
axs[0, 0].plot(
    sol_4C["Time [s]"].entries,
    sol_4C["Terminal voltage [V]"].entries,
    "r--",
    alpha=0.5,
    label="4C Full",
)
(cursor_1C,) = axs[0, 0].plot([], [], "bo", markersize=8, label="1C Current")
(cursor_4C,) = axs[0, 0].plot([], [], "ro", markersize=8, label="4C Current")
axs[0, 0].set_ylabel("Voltage [V]")
axs[0, 0].set_xlabel("Time [s]")
axs[0, 0].legend()
axs[0, 0].grid(True)

# Panel 2: Electrolyte Concentration Profile
(line_ce_1C,) = axs[0, 1].plot([], [], "b-", lw=2, label="1C Rate")
(line_ce_4C,) = axs[0, 1].plot([], [], "r-", lw=2, label="4C Rate")
axs[0, 1].set_xlim(x_ce[0], x_ce[-1])
axs[0, 1].set_ylim(
    min(ce_1C.min(), ce_4C.min()) * 0.9, max(ce_1C.max(), ce_4C.max()) * 1.1
)
axs[0, 1].set_xlabel("Normalized Cell Thickness (Anode -> Separator -> Cathode)")
axs[0, 1].set_ylabel("Electrolyte Conc. [mol/m³]")
axs[0, 1].legend()
axs[0, 1].grid(True)

# Panel 3: Anode Solid Concentration Profile vs Radius
(line_cs_n_1C,) = axs[1, 0].plot([], [], "b-", lw=2, label="1C Rate")
(line_cs_n_4C,) = axs[1, 0].plot([], [], "r-", lw=2, label="4C Rate")
axs[1, 0].set_xlim(r_n[0], r_n[-1])
axs[1, 0].set_ylim(
    min(cs_n_1C_r.min(), cs_n_4C_r.min()) * 0.9,
    max(cs_n_1C_r.max(), cs_n_4C_r.max()) * 1.1,
)
axs[1, 0].set_xlabel("Normalized Particle Radius r (Core -> Surface)")
axs[1, 0].set_ylabel("Anode Solid Conc. [mol/m³]")
axs[1, 0].legend()
axs[1, 0].grid(True)

# Panel 4: Cathode Solid Concentration Profile vs Radius
(line_cs_p_1C,) = axs[1, 1].plot([], [], "b-", lw=2, label="1C Rate")
(line_cs_p_4C,) = axs[1, 1].plot([], [], "r-", lw=2, label="4C Rate")
axs[1, 1].set_xlim(r_p[0], r_p[-1])
axs[1, 1].set_ylim(
    min(cs_p_1C_r.min(), cs_p_4C_r.min()) * 0.9,
    max(cs_p_1C_r.max(), cs_p_4C_r.max()) * 1.1,
)
axs[1, 1].set_xlabel("Normalized Particle Radius r (Core -> Surface)")
axs[1, 1].set_ylabel("Cathode Solid Conc. [mol/m³]")
axs[1, 1].legend()
axs[1, 1].grid(True)

plt.tight_layout()


# ==========================================
# 4. ANIMATION UPDATE FUNCTION
# ==========================================
def update(frame):
    # Update Voltage Cursors
    cursor_1C.set_data([t_anim[frame]], [v_1C[frame]])
    cursor_4C.set_data([t_anim[frame]], [v_4C[frame]])

    # Update Electrolyte Concentration
    line_ce_1C.set_data(x_ce, ce_1C[:, frame])
    line_ce_4C.set_data(x_ce, ce_4C[:, frame])

    # Update Particle Profiles
    line_cs_n_1C.set_data(r_n, cs_n_1C_r[:, frame])
    line_cs_n_4C.set_data(r_n, cs_n_4C_r[:, frame])

    line_cs_p_1C.set_data(r_p, cs_p_1C_r[:, frame])
    line_cs_p_4C.set_data(r_p, cs_p_4C_r[:, frame])

    return (
        cursor_1C,
        cursor_4C,
        line_ce_1C,
        line_ce_4C,
        line_cs_n_1C,
        line_cs_n_4C,
        line_cs_p_1C,
        line_cs_p_4C,
    )


# Render Animation
ani = animation.FuncAnimation(
    fig, update, frames=len(t_anim), interval=50, blit=True
)

plt.show()
```
